## 3. AES-256 Encryption Integration

CPDPM combines polynomial transformations with proven AES-256 encryption. Let's examine the AES integration component.

```python
# Initialize AES components for demonstration
backend = default_backend()

def demonstrate_aes_encryption(data: bytes, password: str) -> dict:
    """Demonstrate standard AES-256 encryption for comparison."""
    
    # Generate cryptographic components
    salt = os.urandom(16)
    iv = os.urandom(16)
    
    # Derive key using PBKDF2
    kdf = PBKDF2HMAC(
        algorithm=hashes.SHA256(),
        length=32,  # 256-bit key
        salt=salt,
        iterations=100000,
        backend=backend
    )
    aes_key = kdf.derive(password.encode())
    
    # Apply PKCS7 padding
    padder = PKCS7(128).padder()
    padded_data = padder.update(data) + padder.finalize()
    
    # Encrypt with AES-256-CBC
    cipher = Cipher(algorithms.AES(aes_key), modes.CBC(iv), backend=backend)
    encryptor = cipher.encryptor()
    ciphertext = encryptor.update(padded_data) + encryptor.finalize()
    
    return {
        'ciphertext': base64.b64encode(ciphertext).decode('utf-8'),
        'salt': base64.b64encode(salt).decode('utf-8'),
        'iv': base64.b64encode(iv).decode('utf-8'),
        'algorithm': 'AES-256-CBC'
    }

# Demonstrate AES encryption
test_data = b"Sample healthcare data for AES demonstration"
test_password = "demo only not a real password"

print("🔐 AES-256 Encryption Demonstration")
print(f"Original data: {test_data}")
print(f"Data size: {len(test_data)} bytes")

start_time = time.perf_counter()
aes_result = demonstrate_aes_encryption(test_data, test_password)
aes_time = time.perf_counter() - start_time

print(f"\n⏱️ AES encryption time: {aes_time:.6f} seconds")
print(f"🔐 AES ciphertext (first 100 chars): {aes_result['ciphertext'][:100]}...")
print(f"🧂 Salt: {aes_result['salt']}")
print(f"🎲 IV: {aes_result['iv']}")
print(f"📏 Ciphertext size: {len(aes_result['ciphertext'])} characters")

# Key size analysis
print("\n🔑 AES Key Properties:")
print(f"  Key size: 256 bits")
print(f"  Block size: 128 bits")
print(f"  Salt size: 128 bits")
print(f"  IV size: 128 bits")
print(f"  PBKDF2 iterations: 100,000")
```

## 4. CPDPM Encryption Implementation

Now let's see the complete CPDPM system in action, combining polynomial transformations with AES-256 encryption.

```python
# Initialize CPDPM hybrid encryption system
cpdpm = CPDPMHybrid(prime_modulus=2**31 - 1)
key_manager = CPDPMKeyManager()

print("🚀 CPDPM Hybrid Encryption System Initialized")

# Demonstrate CPDPM encryption process
healthcare_data = b"""
{
  "patient_id": "DEMO-001",
  "timestamp": "2024-01-15T10:30:00Z",
  "vital_signs": {
    "heart_rate": 72,
    "blood_pressure": "120/80",
    "temperature": 98.6,
    "oxygen_saturation": 99
  },
  "medications": ["Lisinopril 10mg", "Metformin 500mg"],
  "diagnosis": "Type 2 Diabetes, Hypertension"
}
"""

# Generate secure password and context
password = key_manager.generate_secure_password(32)
context = key_manager.derive_context_from_healthcare_data(
    patient_id="DEMO-001",
    timestamp="2024-01-15T10:30:00Z", 
    data_type="vital_signs_record"
)

print(f"📊 Healthcare data size: {len(healthcare_data)} bytes")
print(f"🔐 Generated password length: {len(password)} characters")
print(f"🏥 Healthcare context: {context}")

# CPDPM Encryption Process
print("\n🔄 CPDPM Three-Stage Encryption Process:")

start_time = time.perf_counter()

# Stage 1 & 2 & 3: Polynomial transformation + AES encryption + Integrity protection
encrypted_package = cpdpm.encrypt(
    plaintext=healthcare_data,
    password=password,
    context=context,
    polynomial_degree=7  # Higher degree for enhanced security
)

cpdpm_encryption_time = time.perf_counter() - start_time

print(f"✅ CPDPM encryption completed in {cpdpm_encryption_time:.6f} seconds")

# Display encryption results
print(f"\n📦 CPDPM Encrypted Package:")
print(f"  Algorithm: {encrypted_package['metadata']['algorithm']}")
print(f"  Polynomial degree: {encrypted_package['metadata']['polynomial_degree']}")
print(f"  Original size: {encrypted_package['metadata']['plaintext_length']} bytes")
print(f"  Ciphertext size: {len(encrypted_package['ciphertext'])} characters")
print(f"  Context hash: {encrypted_package['metadata']['context_hash'][:16]}...")

# Show polynomial key properties
poly_key_info = encrypted_package['polynomial_key']
print(f"\n🔑 Polynomial Key Properties:")
print(f"  Degree: {poly_key_info['degree']}")
print(f"  Coefficients count: {len(poly_key_info['coefficients'])}")
print(f"  Max coefficient: {max(poly_key_info['coefficients']):,}")
print(f"  Prime modulus: {poly_key_info['prime_modulus']:,}")

# Security strength estimation
security_analysis = key_manager.estimate_security_strength(
    polynomial_degree=poly_key_info['degree'],
    password_entropy=len(password) * 6  # Rough entropy estimate
)

print(f"\n🛡️ Security Strength Analysis:")
for metric, value in security_analysis.items():
    if isinstance(value, float):
        print(f"  {metric.replace('_', ' ').title()}: {value:.1f}")
    else:
        print(f"  {metric.replace('_', ' ').title()}: {value}")
```

## 5. CPDPM Decryption Implementation

Let's demonstrate the complete CPDPM decryption process and verify data integrity.

```python
# CPDPM Decryption Process
print("🔓 CPDPM Three-Stage Decryption Process:")
print("  Stage 1: Integrity verification")
print("  Stage 2: AES-256 decryption") 
print("  Stage 3: Inverse polynomial transformation")

start_time = time.perf_counter()

# Decrypt using CPDPM
try:
    decrypted_data = cpdpm.decrypt(encrypted_package, password)
    cpdpm_decryption_time = time.perf_counter() - start_time
    decryption_success = True
except Exception as e:
    print(f"❌ Decryption failed: {e}")
    decryption_success = False
    cpdpm_decryption_time = 0

if decryption_success:
    print(f"✅ CPDPM decryption completed in {cpdpm_decryption_time:.6f} seconds")
    
    # Verify data integrity
    data_integrity = healthcare_data == decrypted_data
    print(f"🔍 Data integrity verified: {data_integrity}")
    
    if data_integrity:
        print(f"📋 Recovered healthcare data ({len(decrypted_data)} bytes):")
        # Pretty print the JSON data
        try:
            healthcare_json = json.loads(decrypted_data.decode('utf-8'))
            print(json.dumps(healthcare_json, indent=2))
        except:
            print(decrypted_data.decode('utf-8', errors='replace'))
    
    # Performance comparison
    print(f"\n⚡ Performance Comparison:")
    print(f"  CPDPM Encryption: {cpdpm_encryption_time:.6f} seconds")
    print(f"  CPDPM Decryption: {cpdpm_decryption_time:.6f} seconds")
    print(f"  AES-256 Encryption: {aes_time:.6f} seconds")
    print(f"  CPDPM overhead: {((cpdpm_encryption_time - aes_time) / aes_time * 100):.1f}%")

# Test with wrong password
print(f"\n🚫 Testing with incorrect password...")
try:
    wrong_decryption = cpdpm.decrypt(encrypted_package, "WrongPassword123!")
    print("❌ Security issue: Wrong password accepted!")
except Exception as e:
    print("✅ Security confirmed: Wrong password rejected")
    print(f"   Error type: {type(e).__name__}")

# Test data tampering resistance
print(f"\n🛡️ Testing tampering resistance...")
tampered_package = encrypted_package.copy()
tampered_ciphertext = base64.b64decode(tampered_package['ciphertext'])
tampered_ciphertext = tampered_ciphertext[:-1] + bytes([tampered_ciphertext[-1] ^ 1])
tampered_package['ciphertext'] = base64.b64encode(tampered_ciphertext).decode('utf-8')

try:
    tampered_decryption = cpdpm.decrypt(tampered_package, password)
    print("❌ Security issue: Tampered data accepted!")
except Exception as e:
    print("✅ Security confirmed: Tampered data detected and rejected")
    print(f"   Error type: {type(e).__name__}")

# Validate CPDPM round-trip functionality
validation_result = cpdpm.validate_encryption(healthcare_data, password, context)
print(f"\n🔄 CPDPM round-trip validation: {'✅ PASSED' if validation_result else '❌ FAILED'}")
```
